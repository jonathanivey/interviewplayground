# Creating a Custom Study or Preset

This notebook builds a `Study` from scratch for a new interview topic,
then shows how to save it as a reusable preset. See
[docs/creating_a_preset.md](../docs/creating_a_preset.md) for the
same workflow as a written guide.

**Requirements:** an `OPENAI_API_KEY` (or another provider configured
per the [README](../README.md#setup)) — this notebook makes several LLM calls.

````python
import os

# os.environ["OPENAI_API_KEY"] = "sk-..."

from interviewplayground import Study
````

## Step 1 — Identify insights

Insights are not LLM-generated: review the published study (or other source
material) you're basing the preset on and manually write down the key
findings you want simulated participants to be able to produce — the same
process used to build the three built-in presets. Aim for around 15
distinct insights.

````python
study = Study(insights=[
    "How caregivers first took on the caregiving role and whether it was a choice",
    "Daily routines caregivers use to maintain their own mental health",
    "Support networks caregivers rely on, both formal and informal",
    "How caregivers' sense of identity changes over the course of caregiving",
    # ... continue until you have ~15 distinct insights
])

study.insights
````

## Step 2 — Create participants

````python
study.create_participants(2)
````

## Step 3 — Set personas and traits, create blank memories

````python
p = study.participants[0]

p.persona = (
    "Maria is a 52-year-old woman who left her part-time job three years ago to care "
    "for her mother, who has moderate Alzheimer's disease. She is warm and reflective "
    "but carries significant grief about her mother's decline."
)
p.knowledge = "Medium"
p.verbosity = "High"
p.memory = "High"
p.reflexivity = "High"
p.disclosure = "Medium"
p.understanding = "Medium"

p.create_blank_memories(12)
````

Repeat for the second participant:

````python
p2 = study.participants[1]
p2.persona = (
    "James is a 61-year-old man caring for his wife, who has early-onset Alzheimer's. "
    "He is practical and reserved, and tends to downplay his own stress."
)
p2.knowledge = "Medium"
p2.verbosity = "Medium"
p2.memory = "Medium"
p2.reflexivity = "Medium"
p2.disclosure = "Low"
p2.understanding = "Medium"

p2.create_blank_memories(12)
````

## Step 4 — Distribute insights

````python
study.distribute_insights(avg_per_participant=4.0, ensure_all_distributed=True)
````

## Step 5 — Generate background memories

````python
for p in study.participants:
    p.generate_background_memories()
````

## Step 6 — Generate insight memories

````python
for p in study.participants:
    p.generate_insight_memories(study.insights)
````

## Step 7 — Interview a participant

````python
response = study.participants[0].ask(
    "Can you tell me a bit about how you ended up in a caregiving role?"
)
print(response)
````

## Add research questions and an interview guide

These let InterviewReportCard (and any AI interviewer you test) know
what the interview should cover.

````python
study.research_questions = [
    "How do caregivers first take on the caregiving role, and is it a choice?",
    "What support networks do caregivers rely on?",
]

study.interview_guide = [
    {
        "topic": "Becoming a Caregiver",
        "subtopics": [
            "The circumstances that led to taking on the caregiving role",
            "Whether the role felt like a choice or an obligation",
        ],
    },
    {
        "topic": "Support Networks",
        "subtopics": [
            "Family members or friends who provide support",
            "Use of formal support services or caregiver groups",
        ],
    },
]
````

## Save and register as a preset

Once the `Study` is fully populated, save it and move it into the
package's `presets/` directory so `load_preset()` can find it:

```python
study.save("caregiver_wellbeing.json")
# then move caregiver_wellbeing.json into src/interviewplayground/presets/
# and add "caregiver_wellbeing": _PRESET_DIR / "caregiver_wellbeing.json"
# to _PRESETS in src/interviewplayground/presets/__init__.py
```

This cell just saves it to the current directory so you can inspect the file:

````python
study.save("caregiver_wellbeing.json")
````
